**sqlite3**


A built-in Python module that provides a lightweight, file-based SQL database engine. It allows you to store, query, and manage relational data directly inside a local file (like example.db) without needing a separate server.

```python
#sqllite
import sqlite3
```

sqlite3.connect('database_name.db')

Establishes a connection object to the specified SQLite database file. If the file doesn't exist yet, SQLite automatically creates it.

```python
#connect to sqlite database
connection = sqlite3.connect('example.db')
connection
```

connection.cursor()

Creates a cursor object, which acts as a control structure or "pointer" used to execute SQL queries and traverse the records returned from the database.

```python
cursor = connection.cursor()
```

cursor.execute(sql_query)

Sends a single SQL statement (such as CREATE TABLE, INSERT, UPDATE, DELETE, or SELECT) to the database for processing.

```python
#create table
cursor.execute('''
Create Table If Not Exists employees(
                                      id Integer Primary Key,
                                      name Text Not Null,
                                      Age Integer,
                                      Department Text
                                      )
                                     ''' )
#commit changes
connection.commit()
```

connection.commit()

Saves (commits) all current changes/transactions made to the database since the last commit. Without calling commit(), modifications like INSERT, UPDATE, or DELETE won't be permanently saved.

```python
#Insert the data in sqlite table
cursor.execute('''
Insert Into employees(name,age,department)
Values('shivu',23,'Data scientist')
''')
cursor.execute('''
Insert Into employees(name,age,department)
Values('ram',25,'space scientist')
''')
cursor.execute('''
Insert Into employees(name,age,department)
Values('charlie',35,'finance scientist')
''')
#commit changes
connection.commit()
```

cursor.fetchall()

Fetches all remaining rows from a query result as a Python list of tuples, where each tuple represents a single table row.

```python
#querying the data from sqllite3
cursor.execute('Select * From employees')
rows = cursor.fetchall()

#print the queried data
for row in rows:
  print(rows)
```

```python
#update the data into table
cursor.execute('''
UPDATE employees
set Age = 24
where name = "shivu"
''')
connection.commit()
```

```python
#querying the data from sqllite3
cursor.execute('Select * From employees')
rows = cursor.fetchall()

#print the queried data
for row in rows:
  print(rows)
```

```python
#delete the data from the table
cursor.execute('''
DELETE FROM employees
WHERE name = 'charlie'
''')
connection.commit()
```
