```python
#temp conversion
def convert_temp(temp,unit):
  if unit=='c':
    return temp *9/5+32
  elif unit=='f':
    return (temp-32) *9/5
print(convert_temp(25,'c'))
print(convert_temp(77,'f'))
```

```python
def is_palindrome(s):
  return s==s[::-1]
print(is_palindrome("level"))
print(is_palindrome("woman"))
```

```python
def factorial(n):
  if n==0:
    return 1
  else:
    return n*factorial(n-1)
print(factorial(6))
```

```python
def count_word_frequency(file_path):
  word_count={}
  try:
    with open(file_path,'r') as file:
      for line in file:
          words_in_line = line.split()
          for word in words_in_line:
             word = word.lower().strip('.?/!:;')
             word_count[word]= word_count.get(word,0)+1
    return word_count
  except FileNotFoundError:
    print(f"Error: The file '{file_path}' was not found.")
    return {}

file_path = 'sample.txt'

# Create a sample file for demonstration
with open(file_path, 'w') as f:
    f.write("This is a sample text file.\n")
    f.write("This file contains sample text.\n")
    f.write("Sample text is useful.")

word_frequency_result = count_word_frequency(file_path)
print(word_frequency_result)
```

```python
def add_to_dict(a,b=None):
  if b is None:
    b={}
  b[a]=a**2
  return b
print(add_to_dict(2))
print(add_to_dict(3,{1:1}))
```

```python
def average(lst):
  try:
    return sum(lst)/len(lst)
  except ZeroDivisionError:
    return None
print(average([1,2,3,4,5]))
print(average([]))
```

```python
def write_to_file(lst, filename):
    try:
        with open(filename, 'w') as f:
            for num in lst:
                f.write(f"{num}\n")
    except IOError as e:
        print(f"An error occurred: {e}")

# Test
write_to_file([1, 2, 3, 4, 5], 'output.txt')
```

```python
def make_counter():
    count = 0
    def counter():
        nonlocal count
        count += 1
        return count
    return counter

counter = make_counter()
counter()  # 1
counter()
counter()
```

```python
def perimeter_rectangle(lenght,breadth):
  return int(2*(lenght+breadth))
print(perimeter_rectangle(2,6))
```

```python
import math
def area_of_circle(radius):
  return int(math.pi * radius**2)
print(area_of_circle(2))
```

```python
import math
def area_of_triangle(breadth,height):
  return int(1/2*breadth*height)
print(area_of_triangle(4,6))
```
