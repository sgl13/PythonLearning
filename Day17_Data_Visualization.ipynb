**sales Data visualization**

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("/content/sales_data.csv")
df.head()
```

```python
#plot total sale by product
df.info()
```

```python
#total sales by product
total_sales_by_product = df.groupby('Product_Category')['Sales_Amount'].sum()
print(total_sales_by_product)
```

```python
total_sales_by_product.plot(kind = 'bar',color = 'teal')
```

```python
sales_trend = df.groupby('Sale_Date')['Sales_Amount'].sum().reset_index()
plt.plot(sales_trend['Sale_Date'],sales_trend['Sales_Amount'])
```

```python
import seaborn as sns
#seaborn
#basic plot
tips = sns.load_dataset('tips')
tips
```

```python
#create scatterplot
import matplotlib.pyplot as plt
sns.scatterplot(x='total_bill',y='tip',data=tips)
plt.title("scatterplot of total bill vs Tip")
plt.show()
```

```python
sns.lineplot(x='size',y='total_bill',data=tips)
plt.title("lineplot of total bill vs size")
plt.show()
```

```python
#categorical plot
sns.barplot(x='day',y='total_bill',data=tips)
plt.title("bar plot of total bill by day")
plt.show()
```

```python
#box plot
sns.boxplot(x='day',y='total_bill',data=tips)
plt.title("box plot of total bill by day")
plt.show()
```

```python
#violin plot
sns.violinplot(x='day',y='total_bill',data=tips)
plt.title("violin plot of total bill by day")
plt.show()
```

```python
#histogram
sns.histplot(tips['total_bill'],bins=10,kde= True)
plt.title("histogram of total bill")
plt.show()
```

```python
#kde plot
sns.kdeplot(tips['total_bill'],fill=True)
plt.title("kde plot of total bill")
plt.show()
```

```python
#pair plot used for relationship between every feature
sns.pairplot(tips)
plt.show()
```

```python
#heatmap
corr = tips[['total_bill','tip','size']].corr()
corr
```

```python
sns.heatmap(corr,annot=True,cmap='coolwarm')
plt.show()
```

```python
#plot total sales by product
plt.figure(figsize=(10,6))
sns.barplot(x='Region',y='Sales_Amount',data=df,estimator=sum)
plt.title("total sales by Region")
plt.show()
```

```python
#plot total sales by product
plt.figure(figsize=(10,6))
sns.barplot(x='Product_Category',y='Sales_Amount',data=df,estimator=sum)
plt.title("total sales by product")
plt.show()
```
