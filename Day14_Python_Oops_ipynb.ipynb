**Constructor (__init__) & Instance Methods**

The __init__ method initializes an object's state, while instance methods define behaviors using self

```python
class BankAccount:
  def __init__(self,owner,balance=0): #constructor
    self.owner = owner
    self.balance = balance

  def deposit(self,amount): #methods
    self.balance += amount
    print(f"{amount} is deposited. New balance is {self.balance}")

  def withdraw(self,amount):
    if amount > self.balance:
      print("Insufficient balance")
    else:
      self.balance -= amount
      print(f"{amount} is withdrawn. New balance is {self.balance}")

  def get_balance(self):
    return self.balance

account = BankAccount("shivu",5000) #obj create account
print(account.balance)
account.deposit(100) # call methods
account.withdraw(300)
print(account.get_balance())
```

**Inheritance**

Inheritance allows a subclass to inherit attributes and methods from one or more parent classes.

Single Inheritance with super():
Extends a parent class's behavior and inherits its initializations.

```python
class car:
  def __init__(self,windows,doors,enginetype):
    self.windows = windows
    self.doors = doors
    self.enginetype = enginetype
  def drive(self):
    print(f"the person will drive the {self.enginetype} car")


car1 = car(4,5,"petrol")
car1.drive()

class Tesla(car):
  def __init__(self,windows,doors,enginetype,is_selfdriving):
    super().__init__(windows,doors,enginetype)
    self.is_selfdriving = is_selfdriving
  def selfdriving(self):
    print(f"tesla supports self driving : {self.is_selfdriving}")
tesla1 = Tesla(4,5,"electric",True)
tesla1.selfdriving()
```

**Multiple Inheritance**

Inheriting from multiple base classes simultaneously:

```python
class animal:
  def __init__(self,name):
    self.name = name
  def speak(self):
    print("subclass must implement this method")
  #base class 2
class pet:
  def __init__(self,owner):
    self.owner = owner
# derived class
class Dog(animal,pet):
  def __init__(self,name,owner):
    animal.__init__(self,name)
    pet.__init__(self,owner)
  def speak(self):
    return f"{self.name} say woof"
#create obj
dog = Dog("buddy","shivu")
print(dog.speak())
print(f"owner:{dog.owner}")
```

**Polymorphism & Method Overriding**

Polymorphism allows different classes to share the same method name, executing class-specific behavior.

```python
#polymorphism
class animal():
  def speak(self):
    return "sound of the animal"
#derived class 1
class dog(animal):
  def speak(self):
    return "woof"
#derived class 2
class cat(animal):
  def speak(self):
    return "meow!"

dog = dog()
cat = cat()

print(dog.speak())
print(cat.speak())
```

```python
class Shape:
    def area(self):
        return "the area of figure"


class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius


def print_area(shape):
    print(f"The area is {shape.area()}")


r = Rectangle(4, 5)
c = Circle(3)

print_area(r)
print_area(c)
```
