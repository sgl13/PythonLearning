**pickling**

python usesd to save objeect like (list,dictionary,set,tuple,models..etc ) into a file

pickling converts a python object in bytes(binary forms)-->> serialization

unpickling converts the byte stream in to original python object-->> de serialization

 save data --->> byte format --->close program --->load databack to original



```python
import pickle
```

```python
data = {"name":"shivu","age":23,"skills":['python','np','ML','pd']}
print(data)
```

```python
#serialization #wb mean write binary
with open("data.pkl","wb") as file:
  pickle.dump(data,file)
print("data saved or pickled successfully")
#file = open("data.pkl","wb")
```

```python
#deserialization :- convert the pickle file in to original file
with open("data.pkl","rb") as file:
  load_data = pickle.load(file)
print(load_data)
```

```python
fruits = ['apple','banana','watermelon','pineapple']
for i in fruits:
  print(i)
```

```python
i = iter(fruits)
print(next(i))
print(next(i))
print(next(i))
print(next(i))
```

```python
_i_ = iter(fruits)
print(next(_i_))
```
