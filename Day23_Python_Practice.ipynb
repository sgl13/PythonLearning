```python
import math
def area_of_circle(radius):
  return int(math.pi * radius**2)
print(area_of_circle(2))
```

```python
def area_of_triangle(breadth,height):
  return int(1/2*breadth*height)
print(area_of_triangle(4,6))
```

```python
def perimeter_triangle(length,breadth):
  return int(1/2*length*breadth)
print(perimeter_triangle(6,8))
```

```python
def factorial(num):
  if num == 0:
    return 1
  else:
    return num * factorial(num-1)
print(factorial(5))
```

```python
def sum_of_two(num1, num2):
  return int(num1 + num2)
print(sum_of_two(2,4))
```

```python
def caluclate_lift_round(n,capacity):
  return math.ceil(n/capacity)
print(caluclate_lift_round(10,3))
print(caluclate_lift_round(7,4))
```

```python
def calculate_y(slope, intercept, x):
    return slope*x + intercept
print(calculate_y(2, 3, 4))
print(calculate_y(1.5,-2, 2))
```

```python
def sum_list(*args):
  return int(sum(*args))
print(sum_list([1,2,3,4,5]))
```

```python
def find_largest(*args):
    return max(args)

print(find_largest(3,8,2,10,5))

print(find_largest(-5, -10, -2, -1, -7))
```

```python
def remove_duplicates(lst):
    return list(set(lst))
print(remove_duplicates([1,2,2,3,4,4,5]))
print(remove_duplicates([4,5,5,4,6,7]))
```

```python
def check_unique(lst):
    return len(lst)==len(set(lst))
print(check_unique([1,2,3,4,5]))
print(check_unique([1,2,3,3,4,5]))
```

```python
def reverse_list(lst):
    result = []
    for i in range(len(lst) - 1, -1, -1):
        result.append(lst[i])
    return result

print(reverse_list([1, 2, 3, 4, 5]))
```

```python
def count_even_odd(lst):
    even = len([i for i in lst if i % 2 == 0])
    odd = len([i for i in lst if i % 2 != 0])
    return even, odd
print(count_even_odd([1,2,3,4,5]))
```

```python
def count_even_odd(lst):
    even = 0
    odd = 0
    for i in lst:
        if i % 2 == 0:
            even += 1
        else:
            odd += 1
    return even, odd

print(count_even_odd([1, 2, 3, 4, 5]))
```
