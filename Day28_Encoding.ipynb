## Target Guided Ordinal Encoding
It is a technique used to encode categorical variables based on their relationship with the target variable. This encoding technique is useful when we have a categorical variable with a large number of unique categories, and we want to use this variable as a feature in our machine learning model.

In Target Guided Ordinal Encoding, we replace each category in the categorical variable with a numerical value based on the mean or median of the target variable for that category. This creates a monotonic relationship between the categorical variable and the target variable, which can improve the predictive power of our model.

```python
import pandas as pd

# create a sample dataframe with a categorical variable and a target variable
df = pd.DataFrame({
    'city': ['New York', 'London', 'Paris', 'Tokyo', 'New York', 'Paris'],
    'price': [200, 150, 300, 250, 180, 320]
})
```

```python
df
```

```python
mean_price=df.groupby('city')['price'].mean().to_dict()
```

```python
df['city_encoded']=df['city'].map(mean_price)
```

```python
df[['price','city_encoded']]
```

```python
import pandas as pd

# Sample Data with different values per category
df = pd.DataFrame({
    'City': ['NYC', 'Paris', 'NYC', 'Tokyo', 'Paris'],
    'Price': [500, 300, 700, 200, 500]
})

# 1. Calculate the mean 'Price' for each 'City'
# NYC: (500 + 700) / 2 = 600
# Paris: (300 + 500) / 2 = 400
# Tokyo: 200 / 1 = 200
city_means = df.groupby('City')['Price'].mean()

# 2. Map the category means back to the original column
df['City_Encoded'] = df['City'].map(city_means)

print(df)
```

```python

# Sample Data (1 = Purchased, 0 = Not Purchased)
df = pd.DataFrame({
    'Device': ['Mobile', 'Desktop', 'Mobile', 'Mobile', 'Desktop', 'Tablet'],
    'Bought': [1, 0, 0, 1, 1, 0]
})

# 1. Calculate the conversion rate (probability of 1) per device:
# Mobile:  (1 + 0 + 1) / 3 = 0.67
# Desktop: (0 + 1) / 2     = 0.50
# Tablet:  (0) / 1         = 0.00
device_rates = df.groupby('Device')['Bought'].mean()

# 2. Replace categories with their target probability
df['Device_Encoded'] = df['Device'].map(device_rates)

print(df)
```

```python
# Sample Data
df = pd.DataFrame({
    'Fruit': ['Apple', 'Apple', 'Banana', 'Banana', 'Orange'],
    'Sales': [10, 20, 30, 50, 40]
})

# Encode: Replace fruit name with its average sales
# Apple: (10 + 20) / 2 = 15
# Banana: (30 + 50) / 2 = 40
# Orange: 40
df['Fruit_Encoded'] = df['Fruit'].map(df.groupby('Fruit')['Sales'].mean())

print(df)
```

```python
# Sample Data
df = pd.DataFrame({
    'Grade': ['A', 'B', 'A', 'C', 'B'],
    'Score': [90, 80, 100, 70, 84]
})

# Encode: Replace Grade with average Score
# A: (90 + 100) / 2 = 95
# B: (80 + 84) / 2 = 82
# C: 70
df['Grade_Encoded'] = df['Grade'].map(df.groupby('Grade')['Score'].mean())

print(df)
```
